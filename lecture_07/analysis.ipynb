```python
# 3. Compare Spectrograms (N_FFT=512 vs N_FFT=2048)
hop = 160
fig, axes = plt.subplots(2, 1, figsize=(11, 7), sharex=True)

for ax, nfft in zip(axes, [512, 2048]):
    S = np.abs(librosa.stft(fire_audio, n_fft=nfft, hop_length=hop, center=True))
    S_db = librosa.amplitude_to_db(S, ref=np.max)
    im = librosa.display.specshow(S_db, sr=FS, hop_length=hop, x_axis='time', y_axis='linear', ax=ax, cmap='inferno', vmin=-70, vmax=0)
    ax.set(title=f"STFT Spectrogram - Window Length: {nfft} samples ({1000*nfft/FS:.1f} ms) | Hop: {hop} samples", ylim=(0, FS/2))

fig.colorbar(im, ax=axes, label='Magnitude (dB relative to max)')
plt.savefig(FIG / 'task_fire_spectrogram_comparison.png', dpi=140, bbox_inches='tight')
plt.show()
```
